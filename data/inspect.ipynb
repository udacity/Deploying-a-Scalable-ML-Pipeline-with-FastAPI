```python
for col in df.columns:
    if col != col.strip():
        print(f"Column has whitespace: '{col}'")
```

```python
for col in df.select_dtypes(include="object").columns:
    mask = df[col] != df[col].str.strip()

    if mask.any():
        print(f"\nWhitespace found in column: {col}")
        print(df.loc[mask, col].head(10).tolist())
```
